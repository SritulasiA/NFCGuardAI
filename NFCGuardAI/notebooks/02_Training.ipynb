# IQSecureNet training and calibration

Set Jupyter working directory to `notebooks/`. Real NFC dataset must be prepared separately.

```python
from nfcguard.dataset.store import SignalStore
from nfcguard.dataset.splits import heldout
from nfcguard.utils import load_config
from nfcguard.training.trainer import train_experiment
```

```python
store=SignalStore("../data/processed")
cfg=load_config("../configs/reproduction.yaml")
splits=heldout(store.subtypes(),seed=cfg["seed"])
```

```python
result=train_experiment(store,splits,cfg,"../outputs/notebooks/holdout")
print("Accuracy:",result["test"]["accuracy"])
print("F1:",result["test"]["f1"])
```
